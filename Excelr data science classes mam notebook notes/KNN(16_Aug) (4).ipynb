```python
import pandas as pd
df = pd.read_csv("sales.csv",index_col="Unnamed: 0")
df
```

```python
#---------------------------------------------------------
x_cont= df[df.columns[[0,1,2,3,6]]]

from sklearn.preprocessing import StandardScaler
SS = StandardScaler()
SS_X = SS.fit_transform(x_cont)
SS_X = pd.DataFrame(SS_X)
SS_X
SS_X.columns = list(df.columns[[0,1,2,3,6]])
SS_X
```

```python
from sklearn.preprocessing import LabelEncoder
LE = LabelEncoder()
SS_X["ShelveLoc"] = LE.fit_transform(df["ShelveLoc"])
SS_X["Urban"]     = LE.fit_transform(df["Urban"])
SS_X["US"]        = LE.fit_transform(df["US"])
SS_X["high"]      = LE.fit_transform(df["high"])
SS_X["Education"]      = LE.fit_transform(df["Education"])
SS_X.head()
```

```python
X=SS_X
y=df.Price
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
from sklearn.neighbors import KNeighborsRegressor
knn=KNeighborsRegressor(n_neighbors=12)
knn.fit(X_train,y_train)

train_acc=knn.score(X_train,y_train)
test_acc=knn.score(X_test,y_test)

train_acc,test_acc
```

```python
from sklearn.model_selection import GridSearchCV
model3=KNeighborsRegressor()
params={
    "n_neighbors":range(1,41)
}

grid=GridSearchCV(model3,param_grid=params)
grid.fit(X_train,y_train)
```
