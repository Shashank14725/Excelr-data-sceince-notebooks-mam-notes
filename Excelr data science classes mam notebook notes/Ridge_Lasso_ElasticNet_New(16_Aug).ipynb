```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
data=pd.read_csv('mtcars.csv')
```

```python

data.head()
```

```python
X=data.drop('mpg',axis=1)
y=data['mpg']
```

```python
X
```

```python
y
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.30,random_state=0)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
lr=LinearRegression()

model_lr=lr.fit(X_train,y_train)
y_pred=model_lr.predict(X_test)
r2_score(y_pred,y_test)
```

```python
y_tr_pred=model_lr.predict(X_train)
r2_score(y_tr_pred,y_train)
```

```python
model_lr.coef_
```

```python
df_lr=pd.DataFrame({"features":X.columns,"Coefficients":model_lr.coef_})
df_lr
```

```python
plt.bar(x="features",height="Coefficients",data=df_lr)
```

```python
from sklearn.linear_model import Ridge
from sklearn.metrics import r2_score
ridge=Ridge(alpha=20)

model_rd=ridge.fit(X_train,y_train)
y_pred_rd=ridge.predict(X_test)
r2_score(y_pred_rd,y_test)
```

```python
y_tr_pred_rd=ridge.predict(X_train)
r2_score(y_tr_pred_rd,y_train)
```

```python
df_rd=pd.DataFrame({"features":X.columns,"Coefficients":model_rd.coef_})
df_rd
```

```python
plt.bar(x="features",height="Coefficients",data=df_rd)
```

```python
from sklearn.linear_model import Lasso
from sklearn.metrics import r2_score
lasso=Lasso(alpha=0.5)

model_ls=lasso.fit(X_train,y_train)
y_pred_ls=lasso.predict(X_test)
r2_score(y_pred_ls,y_test)
```

```python
y_tr_pred_rd=lasso.predict(X_train)
r2_score(y_tr_pred_rd,y_train)
```

```python
df_ls=pd.DataFrame({"features":X.columns,"Coefficients":model_ls.coef_})
plt.bar(x="features",height="Coefficients",data=df_ls)
```

```python
from sklearn.linear_model import ElasticNet
from sklearn.metrics import r2_score
elastic=ElasticNet(alpha=0.5,l1_ratio=0.5)

model_en=elastic.fit(X_train,y_train)
y_pred_en=elastic.predict(X_test)
r2_score(y_pred_en,y_test)
```

```python
df_en=pd.DataFrame({"features":X.columns,"Coefficients":model_en.coef_})
plt.bar(x="features",height="Coefficients",data=df_en)
```
