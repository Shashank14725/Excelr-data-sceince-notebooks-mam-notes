```python
# Set mathematical methods:
# Union: returns elements from both sets.
a = {1,2,3}
b = {3,4,5}
print(a.union(b))
```

```python
# intersection() : It will return common elements

print(a.intersection(b))
```

```python
# difference() : returns elemnts present in first set only.

print(a.difference(b))
```

```python
# symmetric_difference(): returns elements not common 
# from both sets

print(a.symmetric_difference(b))
```

```python
# Real worlds examples of set.
# Remove duplicates from record.

names = ["Amit","Neha","Amit"]
unique_names = set(names)
unique_names

# Common students in two classes

classA = {"Ravi","Amol"}
classB = {"Amol","Neha"}
print(classA.intersection(classB))
```

```python
# Conditional Statements: In python, control structures are used 
# the flow of execution based on conditions.
# The most common control structures are 
# conditional statements like if, if-else, if elif-else and nested if.
```

```python
# if statements: The if statement executes a block of 
# code only when condition is True
age = 18
if age > 18:
    print("You are eligible to vote")
```

```python
number = 4

if number % 2 == 0:
    print("Even number")
else:
    print("odd number")
```

```python
marks = 75

if marks >=90:
    print("Grade A")
elif marks >=75:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
else:
    print("Fail")
```

```python
# Nested conditional statements:

age = 17
has_id = True

if age >= 18:
    if has_id:
        print("You can vote!")
    else:
        print("Id is required")
else:
    print("Underage")   
```
