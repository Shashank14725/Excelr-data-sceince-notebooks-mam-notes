
## Summarization (Encoder–Decoder: T5-small)

T5 (`t5-small`) typically ships TensorFlow weights, so we can load it directly via `pipeline` with `framework="tf"`.


```python

# -------------------------------
# SUMMARIZATION — FULLY COMMENTED
# -------------------------------

from transformers import AutoTokenizer, TFT5ForConditionalGeneration, pipeline

# 1) Load tokenizer (framework-agnostic)
t5_id = "t5-small"
t5_tok = AutoTokenizer.from_pretrained(t5_id)

# 2) Load TF model by converting from PyTorch weights on the fly
t5_tf = TFT5ForConditionalGeneration.from_pretrained(
    t5_id,
    from_pt=True   # <-- crucial: convert PT -> TF to avoid dtype errors
)

# 3) Build a TF pipeline using the loaded objects
summarizer = pipeline(
    task="summarization",
    model=t5_tf,
    tokenizer=t5_tok,
    framework="tf"
)

# 4) Test
article = (
    "Transformers use self-attention to understand relationships between words. "
    "This lets them capture long-range context and generate coherent outputs. "
    "They power modern NLP tasks like chatbots, translation, and summarization."
)
summary = summarizer(article, max_length=40, min_length=10, do_sample=False)
print("Original:", article)
print("Summary :", summary[0]["summary_text"])



```
