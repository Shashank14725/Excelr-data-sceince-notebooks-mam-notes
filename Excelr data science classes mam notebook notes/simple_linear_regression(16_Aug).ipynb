```python
import pandas as pd
data=pd.read_csv('experience_vs_salary.csv')
```

```python
data
```

```python
import matplotlib.pyplot as plt
plt.scatter(x='Experience (Years)',y='Salary ($)',data=data)
```

```python
X=data[['Experience (Years)']]
Y=data['Salary ($)']
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()   #creating an instance for the object
```

```python
lr.fit(X,Y)
```

```python
lr.predict(X)
```

```python
lr.intercept_
```

```python
lr.coef_
```

```python
lr.score(X,Y)
```

```python
import seaborn as sns
sns.regplot(x='Experience (Years)',y='Salary ($)',data=data,ci=False)
```

```python
#alternative metrics to evaluate model performance
from sklearn.metrics import r2_score
```

```python
y_pred=lr.predict(X)
```

```python
r2_score(Y,y_pred)
```

```python
from sklearn.metrics import mean_squared_error
```

```python
mean_squared_error(Y,y_pred)
```
