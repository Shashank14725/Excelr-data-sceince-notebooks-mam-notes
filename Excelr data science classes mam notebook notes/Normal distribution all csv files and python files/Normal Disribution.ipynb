```python
import pandas as pd
import numpy as np
```

```python
beml_df = pd.read_csv("BEML.csv")
beml_df
```

```python
glaxo_df = pd.read_csv("GLAXO.csv")
glaxo_df
```

```python
beml_df.info()
```

```python
glaxo_df.info()
```

```python
beml_df = beml_df[['Date', 'Close']]
glaxo_df = glaxo_df[['Date', 'Close']]
```

```python
beml_df
```

```python
'''The DataFrames have a date column, so we can
create a DatetimeIndex index from this column Date. It will ensure that the rows are sorted by time in
ascending order.'''
glaxo_df = glaxo_df.set_index(pd.DatetimeIndex(glaxo_df['Date']))
beml_df = beml_df.set_index(pd.DatetimeIndex(beml_df['Date']))
```

```python
beml_df
```

```python
import matplotlib.pyplot as plt
import seaborn as sn
%matplotlib inline
plt.plot(glaxo_df.Close) #For line graph, use plt.plot(_)
plt.xlabel('Time')
plt.ylabel('Close Price')
plt.show()
```

```python
plt.plot(beml_df.Close)
plt.xlabel('Time')
plt.ylabel('Close')
plt.show()
```

```python
glaxo_df['gain'] = glaxo_df.Close.pct_change(periods = 1)
beml_df['gain'] = beml_df.Close.pct_change(periods = 1)
```

```python
glaxo_df.head()
```

```python
#drop first row since it is NaN
glaxo_df = glaxo_df.dropna()
beml_df = beml_df.dropna()
```

```python
#Plot the gains
# plt.figure(figsize = (8, 6));
plt.plot(glaxo_df.index, glaxo_df.gain)
plt.xlabel('Time')
plt.ylabel('gain')
```

```python
print('Mean:', round(glaxo_df.gain.mean(), 4))    #round() for getting only certain values after decimal
print('Standard Deviation: ', round(glaxo_df.gain.std(), 4))
```

```python
print('Mean: ', round(beml_df.gain.mean(), 4))    #round() for getting only certain values after decimal
print('Standard Deviation: ', round(beml_df.gain.std(), 4))
```
