```python
# Print all even and odd numbers between given range using For Loop
print('Even Numbers:')
for i in range(1, 51):
  if i % 2 == 0:
    print(i, end=" ")

print(" ") # new line
print('Odd Numbers:')
for i in range(1,51):
  if i%2!=0: # != not equal to
    print(i, end=" ")
```
