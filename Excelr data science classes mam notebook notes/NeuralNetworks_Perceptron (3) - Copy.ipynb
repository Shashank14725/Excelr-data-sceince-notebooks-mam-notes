```python
#===================================================================
import tensorflow as tf
print(tf.__version__)

import datetime
print(f"Notebook last run (end-to-end): {datetime.datetime.now()}")
```

```python
# create a data
import numpy as np
import matplotlib.pyplot as plt

plt.figure(figsize=(5, 2))
# Create features (using tensors)
X = tf.constant([-7.0, -4.0, -1.0, 2.0, 5.0, 8.0, 11.0, 14.0])

# Create labels (using tensors)
y = tf.constant([3.0, 6.0, 9.0, 12.0, 15.0, 18.0, 21.0, 24.0])

# Visualize it
plt.scatter(X, y);
```

```python
# random state
tf.random.set_seed(42)

# create a model using the sequential API
model = tf.keras.Sequential([tf.keras.layers.Dense(1)])

# compile the model
model.compile(loss=tf.keras.losses.mse,
              optimizer = tf.keras.optimizers.SGD(),
              metrics=["mse"])

# changing the row to column dimension
tf.expand_dims(X, axis=-1)
```

```python
#fit the model
model.fit(tf.expand_dims(X, axis=-1), y , epochs=20)
```

```python
mse_values = model.history.history['mse']
mse_values
```

```python
Y_pred1 = model.predict([X])
```

```python
Y_pred1
```

```python
# scatter plot between acutual vs predictions
import matplotlib.pyplot as plt
plt.scatter(X,y)
plt.plot(X,y,color='red')

plt.scatter(X,Y_pred1)
plt.plot(X,Y_pred1,color='blue')

plt.ylabel("y and y predictions")
plt.xlabel("X")
plt.title("Actual vs predictions")
plt.show()
```

```python
# version 2
# random state
tf.random.set_seed(42)

# create a model using the sequential API
model = tf.keras.Sequential([tf.keras.layers.Dense(1)])

# compile the model
model.compile(loss=tf.keras.losses.mse,
              optimizer = tf.keras.optimizers.SGD(),
              metrics=["mse"])

# changing the row to column dimension
tf.expand_dims(X, axis=-1)

#fit the model
model.fit(tf.expand_dims(X, axis=-1), y , epochs=120)

mse_values = model.history.history['mse']

Y_pred2 = model.predict([X])
```

```python
# scatter plot between acutual vs predictions
import matplotlib.pyplot as plt
plt.scatter(X,y)
plt.plot(X,y,color='red')

plt.scatter(X,Y_pred1)
plt.plot(X,Y_pred1,color='blue')

plt.scatter(X,Y_pred2)
plt.plot(X,Y_pred2,color='green')


plt.ylabel("y and y predictions")
plt.xlabel("X")
plt.title("Actual vs predictions")
plt.show()
```
