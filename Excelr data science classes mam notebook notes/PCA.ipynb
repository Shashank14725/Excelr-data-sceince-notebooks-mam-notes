```python
import pandas as pd
import numpy as np
```

```python
cancer_data=pd.read_csv("breast-cancer-wisconsin-data.csv")
```

```python
cancer_data.head()
```

```python
cancer_data.info()
```

```python
cancer_data.drop("id",axis=1,inplace=True)
```

```python
X=cancer_data.iloc[:,1:]
y=cancer_data.iloc[:,0]
```

```python
X
```

```python
y
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_sc=sc.fit_transform(X)
```

```python
X_sc
```

```python
from sklearn.decomposition import PCA
pca=PCA()
X_transformed=pca.fit_transform(X_sc)
```

```python
X_transformed.shape
```

```python
pca.explained_variance_ratio_       #variance calculated for each principle component
```

```python
pca.components_       #to check on weighted matrix(Eigen vector)
```

```python
variance=np.cumsum(pca.explained_variance_ratio_)*100
   #percentage of variance is measured
```

```python
variance
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,5))
plt.plot(variance)
plt.xticks([i for i in range(0,31)])
plt.show()
```

```python
X_final=X_transformed[:,0:6]
```

```python
X_final
```

```python
X_final.shape
```

```python
plt.scatter(X_final[:,0],X_final[:,1])    #plotting the data using first two princple components
```

```python
y
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()

y_le=le.fit_transform(y)
```

```python
y_le
```

```python
plt.scatter(X_final[:,0],X_final[:,1],c=y_le)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X_final,y_le,test_size=0.3,random_state=43)
```

```python
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
lr.fit(X_train,y_train)
```

```python
y_test_pred=lr.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_test_pred)
```

```python
y_train_pred=lr.predict(X_train)
accuracy_score(y_train,y_train_pred)
```

# t-SNE stands for t-Distributed Stochastic Neighbor Embedding.
It is a dimensionality reduction technique, mainly used for visualizing high-dimensional data in 2D or 3D.

Although people often mention it along with PCA, t-SNE is NOT a type of PCA. Both are dimensionality reduction methods,
### Non-Linear Dimensionality Reduction Technique

```python
#t-sne- t-distributed stochastic neighbor embedding
```

```python
from sklearn.manifold import TSNE
tsne=TSNE(n_components=2)
X_tsne=tsne.fit_transform(X_sc)
```

```python
plt.scatter(X_tsne[:,0],X_tsne[:,1],c=y_le)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X_tsne,y_le,test_size=0.3,random_state=43)
```

```python
lr.fit(X_train,y_train)
```

```python
lr.score(X_test,y_test)
```

```python
lr.score(X_train,y_train)
```
